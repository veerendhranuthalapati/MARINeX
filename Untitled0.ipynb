```python
import requests
from pathlib import Path

url = "https://noaaocm.blob.core.windows.net/ais/csv2/csv2025/ais-2025-01-01.csv.zst"

path = Path("/content/ais-2025-01-01.csv.zst")

r = requests.get(url, stream=True, timeout=300)
r.raise_for_status()

with open(path, "wb") as f:
    for chunk in r.iter_content(chunk_size=1024 * 1024):
        if chunk:
            f.write(chunk)

print(f"Downloaded: {path}")
print(f"Size: {path.stat().st_size / 1024**2:.2f} MB")
import zstandard as zstd

with open(path, "rb") as fh:
    dctx = zstd.ZstdDecompressor()
    with dctx.stream_reader(fh) as reader:
        sample = reader.read(20000)

text = sample.decode("utf-8", errors="replace")
print(text)
```

```python
header = text.splitlines()[0]
print("COLUMNS:")
print(header)
```

```python
import csv
from io import StringIO

rows = list(csv.reader(StringIO(text)))

for row in rows[:5]:
    print(row)
```

```python
columns = [c.strip() for c in rows[0]]

wanted = [
    "MMSI",
    "BaseDateTime",
    "LAT",
    "LON",
    "SOG",
    "COG",
    "Heading",
    "VesselType",
    "IMO",
    "CallSign",
    "VesselName",
]

print("Available columns:")
for c in columns:
    print(" ", c)

print("\nPotential MARINeX fields:")
for c in wanted:
    matches = [x for x in columns if x.lower() == c.lower()]
    print(f"{c}: {matches}")
```

```python
!pip -q install duckdb pyarrow pandas
from pathlib import Path

AIS_ROOT = Path("/content/marinecadastre")
RAW = AIS_ROOT / "raw"
PARQUET = AIS_ROOT / "parquet"

RAW.mkdir(parents=True, exist_ok=True)
PARQUET.mkdir(parents=True, exist_ok=True)

print(AIS_ROOT)
from pathlib import Path
import shutil

src = Path("/content/ais-2025-01-01.csv.zst")
dst = RAW / src.name

if src.exists():
    shutil.move(str(src), str(dst))

print(dst)
```

```python
import duckdb

db = duckdb.connect(str(AIS_ROOT / "ais.duckdb"))

query = f"""
SELECT
    mmsi,
    CAST(base_date_time AS TIMESTAMP) AS timestamp,
    longitude,
    latitude,
    sog,
    cog,
    heading,
    vessel_name,
    imo,
    call_sign,
    vessel_type,
    status,
    length,
    width,
    draft,
    cargo,
    transceiver
FROM read_csv_auto(
    '{dst.as_posix()}',
    compression='zstd',
    header=true
)
LIMIT 10
"""

result = db.execute(query).fetchdf()
result
```

```python
stats = db.execute(f"""
SELECT
    COUNT(*) AS rows,
    COUNT(DISTINCT mmsi) AS vessels,
    MIN(base_date_time) AS first_timestamp,
    MAX(base_date_time) AS last_timestamp,
    MIN(longitude) AS min_lon,
    MAX(longitude) AS max_lon,
    MIN(latitude) AS min_lat,
    MAX(latitude) AS max_lat
FROM read_csv_auto(
    '{dst.as_posix()}',
    compression='zstd',
    header=true
)
""").fetchdf()

stats
```

```python
parquet_path = PARQUET / "ais_2025_01_01.parquet"

db.execute(f"""
COPY (
    SELECT
        CAST(mmsi AS BIGINT) AS mmsi,
        CAST(base_date_time AS TIMESTAMP) AS timestamp,
        CAST(longitude AS DOUBLE) AS longitude,
        CAST(latitude AS DOUBLE) AS latitude,
        CAST(sog AS DOUBLE) AS sog,
        CAST(cog AS DOUBLE) AS cog,
        CAST(heading AS DOUBLE) AS heading,
        vessel_name,
        imo,
        call_sign,
        CAST(vessel_type AS INTEGER) AS vessel_type,
        CAST(status AS INTEGER) AS status,
        CAST(length AS DOUBLE) AS length,
        CAST(width AS DOUBLE) AS width,
        CAST(draft AS DOUBLE) AS draft,
        CAST(cargo AS INTEGER) AS cargo,
        transceiver
    FROM read_csv_auto(
        '{dst.as_posix()}',
        compression='zstd',
        header=true
    )
)
TO '{parquet_path.as_posix()}'
(FORMAT PARQUET, COMPRESSION ZSTD)
""")

print(parquet_path)
```

```python
db.execute(f"""
SELECT
    mmsi,
    COUNT(*) AS positions,
    MIN(timestamp) AS first_seen,
    MAX(timestamp) AS last_seen
FROM read_parquet('{parquet_path.as_posix()}')
GROUP BY mmsi
ORDER BY positions DESC
LIMIT 20
""").fetchdf()
```

```python
candidates = db.execute(f"""
SELECT
    mmsi,
    vessel_name,
    imo,
    vessel_type,
    timestamp,
    longitude,
    latitude,
    sog,
    cog,
    heading
FROM read_parquet('{parquet_path.as_posix()}')
WHERE longitude BETWEEN -74.5 AND -73.5
  AND latitude BETWEEN 39.5 AND 40.5
  AND timestamp BETWEEN
      TIMESTAMP '2025-01-01 00:00:00'
      AND TIMESTAMP '2025-01-01 06:00:00'
ORDER BY timestamp
""").fetchdf()

print("Matching AIS points:", len(candidates))
candidates.head()
```
